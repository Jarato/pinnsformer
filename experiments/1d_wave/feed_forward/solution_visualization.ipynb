```python
from pinnsform import *
from pinnsform.model import PINN, FLS, FLW, FullWavelet
import matplotlib.pyplot as plt
```

```python
device = 'cpu'

BETA = 3.0

def u_ana(x,t):
    pi = np.pi
    return np.sin(pi*x)*np.cos(2*pi*t)+0.5*np.sin(BETA*pi*x)*np.cos(2*BETA*pi*t)

problem_domain = ([0.0, 1.0], [0.0, 1.0])
mesh_shape = (201, 201)
test_mesh, _ = generate_mesh_object(mesh_shape, domain=problem_domain, device=device, full_requires_grad=False, border_requires_grad=False)
analytic_solution = u_ana(test_mesh.part[0].cpu().numpy(), test_mesh.part[1].cpu().numpy())
```

```python
analytic_solution = np.reshape(analytic_solution, mesh_shape)
```

```python
def visualize_solution(data, domain):
    #[0,np.pi*2,1,0]
    plt.figure(figsize=(4,3))
    domain_extend = [domain[1][0], domain[1][1], domain[0][0], domain[0][1]]
    plt.imshow(data, extent=domain_extend, aspect='auto')
    plt.xlabel('t')
    plt.ylabel('x')
    plt.title('Predicted u(x,t)')
    plt.colorbar()
    plt.tight_layout()
    #plt.savefig('./1dreaction_pinnsformer_pred.png')
    plt.show()
```

```python
base_dir = os.getcwd()
experiment_name = "FourModels_TS_LBFGS_200"
data_path = os.path.join(base_dir, 'results', experiment_name, 'run')
```

```python
visualize_solution(analytic_solution, problem_domain)
```

```python
model = PINN(in_dim=2, hidden_dim=512, out_dim=1, num_layer=4).to(device)

for i in range(201):
    print(f"epoch {i}")
    model_path = os.path.join(data_path, 'PINN', 'seed_0', f'model_at_{i}.pth')
    model.load_state_dict(torch.load(model_path))
    predicted_solution = np.reshape(f(model, test_mesh).detach().cpu().numpy(), mesh_shape)
    visualize_solution(predicted_solution, problem_domain)
```

```python
RHO = 5.0

def h(x):
    return np.exp( - (x-np.pi)**2 / (2 * (np.pi/4)**2))

def u_ana(x,t):
    return h(x) * np.exp(RHO*t) / ( h(x) * np.exp(RHO*t) + 1 - h(x))

test_points = (201, 201)
test_mesh, _ = generate_mesh_object(test_points, domain=problem_domain, device=device, full_requires_grad=False, border_requires_grad=False, num_seq_steps=5, seq_step_size=1e-3)
print('TEST MESH')
print(test_mesh.part[0][:,0].shape)
print(test_mesh.part[1][:,0].shape)
print('#'*20)
analytic_solution = u_ana(test_mesh.part[0][:,0].cpu().numpy(), test_mesh.part[1][:,0].cpu().numpy())
```
